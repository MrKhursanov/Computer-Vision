### Body detection in the pedestrian

```python
cap = cv2.VideoCapture('/Users/apple/Desktop/Projects/Computer Vision/videos/walking.mp4')

body_classifier = cv2.CascadeClassifier('/Users/apple/Desktop/Projects/Computer Vision/Haarcascades/haarcascade_fullbody.xml')

ret, frame = cap.read()
imshow('Before detection in pedestrian', frame)
if ret:
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    body_parts = body_classifier.detectMultiScale(gray, 1.2 ,3)
    
    for (x,y,w,h) in body_parts:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (0, 100, 255), 2)

cap.release()
imshow('Body detection in pedestrian', frame)
```

```python
cap = cv2.VideoCapture('/Users/apple/Desktop/Projects/Computer Vision/videos/walking.mp4')

w = int(cap.get(3))

h = int(cap.get(4))

output = cv2.VideoWriter('walking_output.avi', cv2.VideoWriter_fourcc('M','J','P','G'), 30, (w,h))

body_classifier = cv2.CascadeClassifier('/Users/apple/Desktop/Projects/Computer Vision/Haarcascades/haarcascade_fullbody.xml')

while(True):
    ret, frame = cap.read()
    if ret:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        bodies = body_classifier.detectMultiScale(gray, 1.2, 3)

        for (x,y,w,h) in bodies:
            cv2.rectangle(frame, (x,y),(x+w, y+h), (0, 100, 255), 1)
        
        output.write(frame)

    else:
        break

cap.release()
output.release()



```

### Car detection

```python
cap = cv2.VideoCapture('/Users/apple/Desktop/Projects/Computer Vision/videos/cars.mp4')

car_classifier = cv2.CascadeClassifier('/Users/apple/Desktop/Projects/Computer Vision/Haarcascades/haarcascade_car.xml')

ret, frame = cap.read()
imshow('Before detection in pedestrian', frame)
if ret:
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    cars = car_classifier.detectMultiScale(gray, 1.2 , 3)
    
    for (x,y,w,h) in cars:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (0, 100, 255), 1)

cap.release()
imshow('Cars detection in pedestrian', frame)
```

```python
cap = cv2.VideoCapture('/Users/apple/Desktop/Projects/Computer Vision/videos/cars.mp4')

w = int(cap.get(3))

h = int(cap.get(4))

output = cv2.VideoWriter('cars_output2.avi', cv2.VideoWriter_fourcc('M','J','P','G'), 30, (w,h)) # cv2.VideoWriter(filename, fourcc, fps, frame_size)

car_classifier = cv2.CascadeClassifier('/Users/apple/Desktop/Projects/Computer Vision/Haarcascades/haarcascade_car.xml')

while(True):
    ret, frame = cap.read()
    if ret:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        cars = car_classifier.detectMultiScale(gray, 1.2, 3)

        for (x,y,w,h) in cars:
            cv2.rectangle(frame, (x,y),(x+w, y+h), (0, 117, 255), 1)
            cv2.putText(frame, 'Car', (x,y-5), cv2.FONT_HERSHEY_SIMPLEX, 
            1, (0, 117, 255), 1, cv2.LINE_AA)
        
        output.write(frame)

    else:
        break

cap.release()
output.release()
```

### Perspective transformers

```python
scan = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/scan.jpg')
imshow('Scan img', scan,10)

gray = cv2.cvtColor(scan, cv2.COLOR_BGR2GRAY)

_ , th2 = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

imshow('Threshold of scan', th2)

contours, hierarchy = cv2.findContours(th2, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(scan, contours, -1, (0, 255, 0), thickness = 1)

imshow('overlaid using contours', scan)
```
