```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 5):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

butterfly = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/butterfly.jpg')
imshow('butterfly', butterfly, size=5)

```

```python
opencv_img = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/opencv_inv.png',0)
imshow('Original opencv logo', opencv_img)
```

```python
kernel = np.ones((5,5), np.uint8)
imshow('Original', opencv_img)
erosion = cv2.erode(opencv_img, kernel, iterations = 1)
imshow('Erosion', erosion)
dilation = cv2.dilate(opencv_img, kernel, iterations = 1)
imshow('Dilation', dilation)
opening = cv2.morphologyEx(opencv_img, cv2.MORPH_OPEN, kernel)
imshow('Opening', opening)
closing = cv2.morphologyEx(opencv_img, cv2.MORPH_CLOSE, kernel)
imshow('Closing', closing)
```

### **Canny Algorithm** is used to find the edges

```python
londonxmas_gray = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/londonxmas.jpeg',0)
imshow('Original', londonxmas_gray, size=10)
```

```python
canny = cv2.Canny(londonxmas_gray, 50, 120)
imshow("Canny Image", canny, size=10)
canny = cv2.Canny(londonxmas_gray, 10, 200)
imshow("Canny Wide", canny, size=10)
canny = cv2.Canny(londonxmas_gray, 240, 240)
imshow("Canny Narrow", canny, size=10)
canny = cv2.Canny(londonxmas_gray, 60, 110)
imshow("Canny 4", canny, size=10)
```

### **Auto Canny** function 

```python
def autoCanny(image):
    blurred_img = cv2.blur(image, ksize= (5,5))
    med_value = np.median(image)
    lower = int(max(0,0.66*med_value))
    upper = int(min(255,1.33*med_value))
    edges = cv2.Canny(image = image, threshold1 = lower, threshold2 = upper)
    return edges
    
londonxmas_autoCanny = autoCanny(londonxmas_gray)
imshow('Auto Canny', londonxmas_autoCanny, size=10)
```

```python
baboon_gray_scalled = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg', 0)
baboon_autoCanny = autoCanny(baboon_gray_scalled)
imshow('Baboon autoCanny', baboon_autoCanny, size=5) 
```
