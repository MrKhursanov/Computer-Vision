```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 5):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

butterfly = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/butterfly.jpg')
imshow('butterfly', butterfly, size=5)

```

```python
opencv_img = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/opencv_inv.png',0)
imshow('Original opencv logo', opencv_img)
```

```python
kernel = np.ones((5,5), np.uint8)
imshow('Original', opencv_img)
erosion = cv2.erode(opencv_img, kernel, iterations = 1)
imshow('Erosion', erosion)
dilation = cv2.dilate(opencv_img, kernel, iterations = 1)
imshow('Dilation', dilation)
opening = cv2.morphologyEx(opencv_img, cv2.MORPH_OPEN, kernel)
imshow('Opening', opening)
closing = cv2.morphologyEx(opencv_img, cv2.MORPH_CLOSE, kernel)
imshow('Closing', closing)
```
