```python
# Our Setup, Import Libaries, Create our Imshow Function and Download our Images
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 5):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

baboon = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg')
```

### Blurring using convolutions

```python
kernel_3x3 = np.ones((3,3), np.float32)/9
kernel_7x7 = np.ones((7,7), np.float32)/49
print(kernel_3x3)
print(kernel_7x7)

img_3x3blured = cv2.filter2D(baboon, -1, kernel_3x3)
img_7x7blured= cv2.filter2D(baboon, -1, kernel_7x7)
imshow('Original', baboon)
imshow('Blured 3x3 Image', img_3x3blured)
imshow('Blured 7x7 Image', img_7x7blured)
```

```python
img1 = cv2.blur(baboon, (5,5))
img2 = cv2.GaussianBlur(baboon, (5,5), 0)
img3 = cv2.medianBlur(baboon, 5)

imshow('Original', baboon)

imshow('Average blur', img1)

imshow('Gaussian Blur', img2)

imshow('medianBlur', img3)
```

### Sharpening an image

```python
kernel_sharpening = np.array([
    [-1, -1, -1],
    [-1, 9, -1],
    [-1, -1, -1]
])

sharpened_img = cv2.filter2D(baboon, -1, kernel_sharpening)
imshow('Original', baboon)
imshow('Sharpened Image', sharpened_img)
```
