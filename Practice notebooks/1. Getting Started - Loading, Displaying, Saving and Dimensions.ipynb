```python
def imshow(title = '', image= None):
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(f'Display our sample photo. {title}')
    plt.show()
imshow(title = 'What is it? Leave comments.', image = baboon)
```
