```python
# Our Setup, Import Libaries, Create our Imshow Function and Download our Images
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

```python
baboon = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg')
```

```python
imshow('Baboon',baboon, size=5)
```

### cv2.warpAffine just changes the location of spots, where its height and width starts

```python
height, width = baboon.shape[:2]
quarter_h, quarter_w = height/2, width/2
T = np.float32([[1,0,quarter_w],[0,1,quarter_h]])
img_transformation = cv2.warpAffine(baboon, T, (width, height))
imshow('Original', baboon, size=4)
imshow('After transformation', img_transformation, size=5) 


```

```python
print(T)
print(height, width)
print(baboon.shape)
print(img_transformation[511])
```

### cv2.getRotationMatrix2D is used to change or rotates the values of the matrice which is its height and width

```python
h,w = baboon.shape[:2] #here we are taking the height and width of the image
rotation_matrix= cv2.getRotationMatrix2D((w/2,h/2), 90, 1)
rotated_image = cv2.warpAffine(baboon, rotation_matrix, (w, h))
imshow('Original', baboon, size=4)
imshow('Transformed image', img_transformation, size=4)
imshow('Rotated Image', rotated_image, size=4)

```

### Rotation an image using cv2.transpose() class, it will just change or find T of a matrice, it's is easy to rotate the matrice easily, but it gets difficult to change its parameters or values or settings


```python
transposed_img = cv2.transpose(baboon)
imshow('Original', baboon, size=4)
imshow('Transformed image', img_transformation, size=4)
imshow('Rotated Image', rotated_image, size=4)
imshow('Transposed Image', transposed_img, size=4)
```

### cv2.flip(img, 1) this class changes the values of matrices horizontaly

```python
flipped_img = cv2.flip(baboon, 1)
imshow('Original', baboon, size=4)
imshow('Transformed image', img_transformation, size=4)
imshow('Rotated Image', rotated_image, size=4)
imshow('Transposed Image', transposed_img, size=4)
imshow('Flipped Image', flipped_img, size=4)
```
