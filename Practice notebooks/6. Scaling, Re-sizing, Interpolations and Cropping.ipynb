```python
# Our Setup, Import Libaries, Create our Imshow Function and Download our Images
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

### Resizing is used the change the values of the images, it affect the quality of the images, coz we are directly reducing or increasing values of the image or matrice, in this condition we use cv2.resize(img, dsize(output img size), x scale, y scale, interpolation), here interpolation is a method, and there are several types of interpolation method, here the core meaning is like that: we zoom in the image, its quality decreases, coz it predicts the close values to the points, but we zoom out there are almost no changes, coz we are not affecting the quality or values directly.

```python
baboon = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg')

resized_img = cv2.resize(baboon, None, fx=0.1, fy=0.1)
resized_img2 = cv2.resize(baboon, None, fx=2, fy=2, interpolation = cv2.INTER_CUBIC) 
resized_img3 = cv2.resize(baboon, None, fx=2, fy=2, interpolation = cv2.INTER_NEAREST) 
resized_img4 = cv2.resize(baboon, (900,400), interpolation = cv2.INTER_AREA) 

imshow('Original Image', baboon, size=5)
imshow('Resized image', resized_img, size=5)
imshow('Resized Cubic image', resized_img2, size=5) # when we zoom out the quality is getting increased
imshow('Resized Nearest Image', resized_img3, size=5) # when we zoom out the quality is getting increased
imshow('Resized Area Image', resized_img4, size=15) # when we zoom in the quality is getting decreased



```

### Image Pyramids

```python
smaller = cv2.pyrDown(baboon)
larger = cv2.pyrUp(smaller)
even_smaller = cv2.pyrDown(smaller)
the_smallest= cv2.pyrDown(even_smaller)

imshow('Original Image', baboon, size=5)
imshow('Resized image', resized_img, size=5)
imshow('Resized Cubic image', resized_img2, size=5) # when we zoom out the quality is getting increased
imshow('Resized Nearest Image', resized_img3, size=5) # when we zoom out the quality is getting increased
imshow('Resized Area Image', resized_img4, size=15) # when we zoom in the quality is getting decreased
imshow('Smaller Image', smaller, size=5)
imshow('Larger than smaller image', larger, size=5)
imshow('Even smaller image', even_smaller, size=5)
imshow('Even smaller image', the_smallest, size=5) # when we zoom in more and more the quality of the image is decreasing
```

### Cropping


```python
h,w = baboon.shape[:2]
start_row, start_col = int(h*0.05),int(w*0.05)
end_row,end_col = int(h*0.9), int(w*0.9)
cropped = baboon[start_row:end_row,start_col:end_col]
copy = baboon.copy()
cv2.rectangle(copy,(start_col, start_row),(end_col,end_row),(0,255,255,),5)

imshow('Original', baboon, size=5)
imshow('Cropped Image', cropped, size=5)
imshow('Rectangle Image', copy, size=5)
```
