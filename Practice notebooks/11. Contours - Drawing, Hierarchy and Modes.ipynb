```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 5):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

butterfly = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/butterfly.jpg')
imshow('butterfly', butterfly, size=5)

```

```python
lp = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/LP.jpg')
imshow('Licence plate', lp, 20)
```

```python
LP_gray = cv2.cvtColor(lp, cv2.COLOR_BGR2GRAY)
imshow('Gray Image', LP_gray, 20)

_ , th2 = cv2.threshold(LP_gray, 127, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
imshow('Threshold', th2, 20)

contours, hiercharhy = cv2.findContours(th2, cv2.RETR_LIST, cv2.CHAIN_APPROX_NONE)
cv2.drawContours(lp, contours, -1, (0, 255, 0), thickness = 2)
imshow('Contours overlaid on original image', lp, 20) # This algorithm is used later when we are detecting the plate numbers and training or testing them, it is pretty usefull when are testing

print('Number of Contours found:', str(len(contours)))
```

```python
body = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/part_of_body.png')
imshow('The parts of the body', body, size=10)
body_gray_scaled = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/images/part_of_body.png', 0)
imshow('The parts of the body in no BGR', body_gray_scaled, size=10)
body_gray = cv2.cvtColor(body, cv2.COLOR_BGR2GRAY)
imshow('The parts of the body in gray', body_gray, size=10) # Both no RGB colors are the same, their functionality is almost the same, but their usage might be different, and it is a little bit later as an advanced usage!
```

```python
_ , thresh2 = cv2.threshold(body_gray,0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
imshow('Threshold of body parts', thresh2, 10)

contours, hiercharhy = cv2.findContours(thresh2, cv2.RETR_LIST, cv2.CHAIN_APPROX_NONE)
cv2.drawContours(body, contours, -1, (0, 255, 0), thickness = 2)
imshow('Contours overlaid on original image', body, 20) # Here it is getting clear that it is better to use or change the color into black or significantly different colors, coz it will get much more easy, when we are analyzing or detecting them!
```
