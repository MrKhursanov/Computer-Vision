```python
# Our Setup, Import Libaries, Create our Imshow Function and Download our Images
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 5):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

baboon = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg')

imshow('Original', baboon, size=5)
```

### Grayscaled image

```python
baboon_gray_scaled = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg', 0)

imshow('Grayscaled Image', baboon_gray_scaled, size=5)
```

### Increasing brightness

```python
baboon_gray_scaled.shape
```

```python
M = np.ones(baboon_gray_scaled.shape, dtype="uint8")*100
print(M)
```

```python
added_M1 = cv2.add(baboon_gray_scaled, M)
imshow('added_M1 using CV2', added_M1, size=5 ) 
added_M2 = baboon_gray_scaled + M
imshow('Just added', added_M2, size=5)
```

```python
h, w = baboon_gray_scaled.shape[:2]
start_row, start_col = int(h*0.18), int(w*0.2)
end_row, end_col = int(h*0.8), int(w*0.8)
cropped = baboon_gray_scaled[start_row : end_row, start_col: end_col]
copy = baboon_gray_scaled.copy()
cv2.rectangle(copy,(start_col, start_row),(end_col,end_row),(2,134,134),5)
imshow('Cropped Image', cropped)
imshow('Rectangle added', copy)
```

```python
added_M3 = M + copy
imshow('Added', added_M3)
print(added_M3)
```

### Decrease the brightness


```python
h, w = baboon_gray_scaled.shape[:2]
start_row, start_col = int(h*0.18), int(w*0.2)
end_row, end_col = int(h*0.8), int(w*0.8)
cropped = baboon_gray_scaled[start_row : end_row, start_col: end_col]
copy = baboon_gray_scaled.copy()
cv2.rectangle(copy,(start_col, start_row),(end_col,end_row),(2,134,134),5)
subtract_M1 = cv2.subtract(baboon_gray_scaled, M)
imshow('Subtracted_M1 using CV2', subtract_M1, size=5 ) 
subtract_M2 = baboon_gray_scaled - M
imshow('Just added', subtract_M2, size=5)
imshow('Cropped Image', cropped)
imshow('Rectangle added', copy)

subtract_M3 = copy - M
imshow('Subtracted', subtract_M3)
print(subtract_M3)
```
