```python
trainset  = torchvision.datasets.mnist('mnist', 
                                       train = True,
                                       download = True,
                                       transform = transform)

testset  = torchvision.datasets.mnist('mnist', 
                                       train = False,
                                       download = True,
                                       transform = transform)
```
