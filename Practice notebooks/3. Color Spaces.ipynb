```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def imshow(title = 'Image', image = None, size = 5):
    h, w = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size*aspect_ratio, size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_RGB2BGR))
    plt.title(title)
    plt.show()
```

```python
baboon = cv2.imread('/Users/apple/Desktop/Projects/Computer Vision/Practice notebooks/baboon.jpg')
```

```python
imshow(title="Baboon", image=baboon, size=10)
```

```python
B, G, R = cv2.split(baboon)
print(B.shape)
print(G.shape)
print(R.shape)
```

```python
imshow('Blue Channel only', B)
```

```python
imshow('Red Channel only', R)
```

```python
imshow('Green Channel only', G)
```

```python
zeros = np.zeros(baboon.shape[:2], dtype='uint8')

imshow('Green Channel only', cv2.merge([zeros, zeros, G]))
```

```python
imshow('Red', cv2.merge([zeros, R, zeros]))
```

```python
imshow('Blue', cv2.merge([B, zeros, zeros]))
```

```python
hsv_image = cv2.cvtColor(baboon, cv2.COLOR_BGR2HSV)
imshow('HSV format:', hsv_image)
```
