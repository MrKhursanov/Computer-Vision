```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
baboon = cv2.imread('/Users/apple/Desktop/Projects/Pytorch/Practice notebooks/baboon.jpg')
```

```python
def imshow(title='Image', image = None, size = 4):
    h, w = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size*aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show

imshow(title = 'Image', image=cv2.imread('/Users/apple/Desktop/Projects/Pytorch/Practice notebooks/baboon.jpg'))
```
